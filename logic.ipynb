## IF

No that you have a variable whihc can tell you if something is true or false, this allows you to make decisions on what the code should do next, using the `if` statement. 

Take note that indented code below an `if` statement must be **4 space characters** to the right of the `if`. Anything more or less will cause an indentation error. 

```python
in_num = int(input("Enter a number: "))
if in_num > 0:
    print("You entered a number.")
    print("It was a positive number")
    print(f"The number was {in_num}.")
```

### ELSE & ELIF

An `if` statement will execute the indented code below it of the test is `True`. If you need to execute a seperate block of code if the test is `False` you can use the `else` statement. You can even futher refine the check by using the `elif` statement (`elif` stands for `else if`).

The same **4 space character** indentation applies to `elif` and `else`.

```python
in_num = int(input("Enter a number: "))
if in_num > 0:
    print("You entered a number.")
    print("It was a positive number")
    print(f"The number was {in_num}.")
elif in_num == 0:
    print("You entered zero")
else:
    print(f"{in_num} is a negative number")
```

### Nested IF

You can also nest `if` statements within each other to help refine your decision making. The same **4 space character** indentation applies to a nested `if`, so the indented code lines for a nested `if` will be four space characters to the right of that nested `if`, as below:

```python
in_num = int(input("Enter a number: "))
if in_num > 0:
    if in_num > 1_000_000:
        print("That was a big number")
    else:
        print(f"Not such a large number")
elif in_num == 0:
    print("You entered zero")
else:
    print(f"{in_num} is a negative number")
```

### OR, NOT, AND

You can construct more intricate test using the `or`, `not`, `and`, qualifiers.

To execute a block of code if one of several conditions are true, use `or` between each test. you can use multiple `or` qualifiers in the same line.

```python
in_char = input("Enter a letter: ")
if in_char == "I" or in_char =="l":   # use of `or`
    print("Do you mean i or l ?")
```

The `not` qualifier is useful to check if an item is in a list as part of an `if` statement:

```python
my_colour = "orange"
colours = ["red", "green", "blue"]
if my_colour not in colours:        # use of `not`
    print(f"{my_colour} is NOT in the list of colours")
else:
    print(f"{my_colour} is in the list of colours")
```

`not` can also be used to reverse the boolean value of a test

```python
in_num = int(input("Enter an age: "))
if in_num > 50:
    print("Age is greaer than 50")
elif not in_num > 50:              # use of `not`
    print("Age is less than 50")
```

### Simplyfying an IF

As you become more confident, you can simplify your `if` test:

```python
in_num = int(input("Enter an age: "))
if in_num >= 13 and in_num <=19:      # Normal `if` format
    print("This is a teenager")
```

```python
in_num = int(input("Enter an age: "))
if 13 <= in_num <= 19:               # Simplified 'if' format
    print("This is a teenager")
```

# FOR

The `for` statement is commonly used to move through each item in a collection of data, for example a list.

```python
customers = ["Susan", "Alex", "Nick", "Mary"]
for value in customers:
     print(value, len(value))
```

If you need to count how many es the for loop has executed, you can add another variable just before the value variable which will be automatically incremented as the `for` moves through the list.

```python
customers = ["Susan", "Alex", "Nick", "Mary"]
for count, value in enumerate(customers):
     print(count, value)
print(f"Loop executed {count+1} times")
```

### Controlling How Many Times a FOR Loops

Rather than using a `list` to loop through you can make a code block a specified number of times by using `range()` to define the limit. 

```python
num = int(input("How many numbers to print"))
for i in range(0, num):
   print(i)
```

## Loop through a DICTIONARY

To look through key/value pairs in a dictionary you must use the `for` loop to extract the key, then use that key to access its value in the normal way, as below

```python
blackbird_dict = {
  "Genus": "Sturnus",
  "Species": "Vulgaris",
  "Length": 22,
  "Min Mass": 58,
}

# A dictionary will only give you its keys when one variable is
# present in the FOR loop
for key in blackbird_dict:
    print(F"{key}: {blackbird_dict[key]}")
```

However, if you want both the key and its value at the same time, use two variables in the for loop:

```python
blackbird_dict = {
  "Genus": "Sturnus",
  "Species": "Vulgaris",
  "Length": 22,
  "Min Mass": 58,
}

# As two variables are present, the key is placed in the first variable
# and its correpsonding value in the second variable
for key, value in blackbird_dict.items():
    print(F"{key}: {value}")
```

```python
### Use FOR and IF to Check an Items Data Type
my_list = [23, "test", 45, 3.45]
print(my_list)

for count, a in enumerate(my_list):
    if type(a) is str:              # Loop list until string is found
        print(f"Found string: {a}") # print the string
        my_list[count] = -1         # use the current loop count to replace the string with -1
print(f"After: {my_list}")
```

### Using range() in a FOR Loop

range() can be useful for splitting data into parts, if each part is the same length.<br>
In the example below,<br>
* the `for` loop is using `range()` to define how many times to loop<br>
* the length of `my_string` is used to control how many loops `for` will make. This is more flexible as the length of `my_string` could change and the code will be able to cope with this.<br>
* rather than looping through items  in sequence, `len()` is told to step (skip/ignore) every 3 loops<br>
* this means that the variable `i` will not loop through values normally such as 1,2,3,4,5,6,7,8,9...... but instead step over 3 loops each time, so on each loop, `i` has the value of 0,3,6,9,12. The maximum number of loops had been set by `len(my_string)` which is 15 (the length of the string) so `for` stops when it reaches 12.

```python
my_string = "...one...two..."

# range(startvalue , stop value, step value)
for i in range(0, len(my_string),3):
    print(i, my_string[i:i+3])
```

## WHILE

A `while` loop is useful to keep repeating a block of code until a specific condition/test is met.

The code below, will keep asking the user for an age until an acceptable value is entered. Negative values and values over 130 will not be accepted.

```python
age = int(input("Enter your age: "))

while age < 0 or age > 130:
    print("Invalid age, please try again")
    age = int(input("Try again: "))

print(f"Valid age entered of : {age}")
```

### BREAK

As a `while` is continuously repeating its block of code, you can `break` out of the loop should a certain condition be met.

```python
x = 1
while x > 0:
    print(x)
    x = x + 1
    if x > 5:
        break
        print("Does code get to this point?")

print("While loop terminated")
```

### CONTINUE

As `while` is executing its block of code, it is possible to stop the current loop, skip the rest of the current block of code and and move straight onto the next loop:  

```python
x = 0
while x >= 0:
    x = x + 1
    if x == 3:
        continue  # If code reached this line, the rest of code below
                  # in this block is skipped over, and the next loop starts
                  # so the elif and print statemetns do not gt executed
                  # when x==3 is true
    elif x == 6:
        break
    print(x)
```
