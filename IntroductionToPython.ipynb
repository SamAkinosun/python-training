We can carry out a calculation and then place the answer straight into a variable, see the example below.

```python
# This is an example of hard coding values, we use the raw values in the calculation
# Not good practice
total = 32 - 6
total
```

However, this is not really a good idea. There is no way to tell what the numbers 32 and 6 represent. When we use numbers directly in a calculation like this , it is called **hardcoding** the values, and something you should always avoid.
It is better practice to place the numbers into a variable, and use the variable in the calculation. This makes the code easier to understand and easier to change values in large calculations where same value is used more than once.

```python
# Using variable names
cost = 32
discount = 6
total = cost - discount
total
```

### Data (Variable) Types

The terms 'Variable types' and 'data types' are often used interchangeably to mean the type of data stored in the variable.

The three main variable (data) types are:
* integer - represented as `int`, example: `1`
* float - represented as `float`, example: `1.2`
* string - represented as `str`, example `"One"`

In other languages you have to define a variable to hold one specific type of data. In python, a variable does not care what data type you place in it. You can use the `type` keyword to discover what type of data is being held in a variable.

```python
# Check to see what data type x is
x = 32
type(x)
```

```python
# Check to see if the variable x is a string type
x = "Hello World"
type(x) is str
```

An alternative method of checking the type of a variable is to use `isinstance()`:

```python
a = 2.3
isinstance(a,float)
```

### Making Numbers Easier to Read

Large numbers can be made easier to read my using an underscore character to separate the digits. Take the example below:

```python
# this number is quite difficult to read
x = 3000000000
x
```

The following can be easier to read and understand:

```python
# Using _ to make large numbers more readable
x = 3_000_000_000
x
```

You can see the output from both previous examples is the same. The underscore is purely cosmetic and has no impact on the data itself.

### Converting Data Types

There are commands you can use to convert one data type to another. these are:
* `int()`
* `float()`
* `str()`

```python
# this example converts x from an int to a float
x = 1        # x Starts as an integer of 1
x = float(x) # Here we place the result of float(x) back into x
x            # The value inside x has been convereted to a float
```

Another example below shows how to converts a string to a float. Strings do not just contain characters, they can contain numbers and special characters. If a string contains numbers, that string cannot be used in a calulation. you first need to convert it to an integer or float, as in the example below.

```python
y = "34.8"
z = float(y)
type(z)
```

Look at the code in the cell below. It attempts to convert the **string** "3.4" to an **integer**. What happens when you run the code.

```python
# Convert a string to an integer, what ahppens?
x = "3.4"
y = int(x)
y
```

Now look at the code below. First the string is converted to a float, then the float is converted to an integer. What is the result.

```python
x = "3.4"
y = float(x)
z = int(y)
z
```

In the previous example the error is caused by int() being unable to convert a string which has a decimal point to an integer. However, float() can convert a string with a decimal point (as long as there are no letters in the string). So we can use this to convert the string to a float, then convert the float to an integer. Note how int() removes the decimal point, resulting in z being 3.

The point of this example is to show that converting one data type to another can present challenges and is something you will have to do on a regular basis in data science and AI.
