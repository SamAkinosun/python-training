## Booleans and comparisons

A **boolean** is one of two values: `True` or `False` (note the capital letters).
Comparing two values produces a boolean.

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `<=` | less than or equal |
| `>` | greater than |
| `>=` | greater than or equal |

A single `=` assigns a value. A double `==` compares. Mixing them up is a classic
beginner mistake.

```python
print(3 == 3)        # True
print(3 == 4)        # False
print(3 != 4)        # True
print("cat" == "Cat")  # False: capitalisation matters
```

## Combining tests

- `and` is True only if **both** sides are True.
- `or` is True if **either** side is True.
- `not` flips a boolean.

```python
age = 25
print(age > 18 and age < 65)   # True: both are true
print(age < 13 or age > 65)    # False: neither is true
print(not age > 18)            # False: age > 18 is True, 'not' flips it
```

## if / elif / else

`if` runs its indented block only when its test is True. `elif` (else-if) checks another
condition, and `else` covers everything left over. The indented lines must be **4 spaces**
in; Jupyter does this for you when you press Enter after a colon.

```python
score = 72

if score >= 90:
    print("Grade A")
elif score >= 70:
    print("Grade B")
elif score >= 50:
    print("Grade C")
else:
    print("Below passing")
```

## Membership tests

`in` and `not in` check whether a value is part of a collection, such as a list of
allowed options. They read almost like English.

```python
colour = "orange"
allowed = ["red", "green", "blue"]

if colour not in allowed:
    print(f"{colour} is not an allowed colour")
else:
    print(f"{colour} is allowed")
```

## Nesting and simplifying

You can put an `if` inside another `if` to ask follow-up questions.

```python
number = 2_500_000

if number > 0:
    if number > 1_000_000:
        print("A large positive number")
    else:
        print("A small positive number")
else:
    print("Zero or negative")
```

When you are checking whether a value sits in a range, Python lets you chain the
comparison, which is shorter and clearer than joining two tests with `and`.

```python
age = 16

# These two checks are equivalent:
print(age >= 13 and age <= 19)
print(13 <= age <= 19)
```
