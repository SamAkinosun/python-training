## Arithmetic operators

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | division | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

```python
print(7 + 2)
print(7 / 2)     # always gives a float
print(7 // 2)    # whole-number division, drops the remainder
print(7 % 2)     # the remainder left over
print(7 ** 2)    # 7 to the power of 2
```

`//` and `%` are more useful than they look. `%` tells you if a number is even
(`n % 2 == 0`), and together they split a total into whole units plus a leftover, for
example converting minutes into hours and minutes.

```python
total_minutes = 145
hours = total_minutes // 60
minutes = total_minutes % 60
print(hours, "hours and", minutes, "minutes")
```

## Order of operations

Python follows normal maths rules: `**` first, then `*` `/` `//` `%`, then `+` `-`.
Use parentheses to make the order explicit and the code readable.

```python
print(3 + 4 * 5)        # 23, because * happens before +
print((3 + 4) * 5)      # 35, because the brackets happen first
```

## Assigning several variables at once

You can assign multiple variables on one line. The values line up with the names by
position.

```python
x, y, z = 1, 2.0, "three"
print(x)
print(y)
print(z)
```

## Readable large numbers

Long numbers are hard to read. You may use underscores as separators; they are purely
cosmetic and do not change the value.

```python
hard_to_read = 3000000000
easy_to_read = 3_000_000_000
print(hard_to_read)
print(easy_to_read)
print(hard_to_read == easy_to_read)   # True: same value
```
