## Variables

A **variable** is a named box that holds one value. The `=` sign puts the value on the
right into the name on the left.

````python
cost = 32
print(cost)
````

You can use a variable in calculations, and you can replace its value at any time.
Putting values into well named variables makes code easier to read and change than
writing raw numbers everywhere.

````python
cost = 32
discount = 6
total = cost - discount
print(total)
````

Names are **case sensitive**: `cost` and `Cost` are two different variables. Use lower
case names with underscores, like `unit_price`, which is the common Python style.

## The core types

Every value has a **type**. The four you will use constantly are:

| Type | Means | Example |
|------|-------|---------|
| `int` | whole number | `42` |
| `float` | number with a decimal point | `3.14` |
| `str` | text (a "string") | `"hello"` |
| `bool` | True or False | `True` |

Python figures out the type from the value; you do not declare it up front.

````python
quantity = 42          # int
price = 3.14           # float
name = "Widget"        # str
in_stock = True        # bool

print(quantity, price, name, in_stock)
````

## Checking a type

Use `type()` to see what a value is, or `isinstance()` to ask a yes/no question about
whether a value is a particular type.

````python
x = 32
print(type(x))

y = "Hello"
print(type(y))

print(isinstance(3.14, float))   # True
print(isinstance(3.14, int))     # False
````

## Converting between types

`int()`, `float()`, and `str()` convert a value from one type to another. This comes up
constantly, for example when a number arrives as text and you need to do maths with it.

````python
text = "34.8"
number = float(text)     # turn the string into a float
print(number, type(number))
````

### A common pitfall

`int()` cannot convert a string that contains a decimal point. This fails:

```python
int("3.4")     # ValueError
```

The fix is to convert the string to a `float` first, then to an `int`. Notice that
turning a float into an int **drops** the decimal part (it does not round).

````python
text = "3.4"
as_float = float(text)   # 3.4
as_int = int(as_float)   # 3  (the .4 is dropped, not rounded)
print(as_float, as_int)
````
