# Loops

## What you'll learn
- Repeating work with a `for` loop
- Generating number sequences with `range()`
- Counting items with `enumerate()`
- Repeating until a condition with `while`
- Stopping or skipping with `break` and `continue`

## The for loop

A `for` loop runs its indented block once for each item in a collection. The loop
variable (`customer` below) takes each value in turn.

```python
customers = ["Susan", "Alex", "Nick", "Mary"]

for customer in customers:
    print(customer, "has", len(customer), "letters")
```

You can loop over a string too; each item is one character.

```python
for letter in "cat":
    print(letter)
```

## range(): looping a set number of times

`range(start, stop)` produces the numbers from `start` up to but **not including**
`stop`. With one argument, `range(n)` goes from 0 to `n - 1`.

```python
for i in range(5):       # 0, 1, 2, 3, 4
    print(i)

print("---")

for i in range(2, 6):    # 2, 3, 4, 5
    print(i)
```

A third argument is the **step**, which lets you skip. This is handy for walking through
data in fixed-size chunks.

```python
for i in range(0, 10, 2):    # 0, 2, 4, 6, 8
    print(i)
```

## enumerate(): item and its position

When you need both the item and its index, `enumerate()` gives you both at once, which is
cleaner than keeping a separate counter.

```python
customers = ["Susan", "Alex", "Nick"]

for index, customer in enumerate(customers):
    print(index, customer)
```

## The while loop

A `while` loop repeats as long as its condition stays True. Use it when you do not know
in advance how many times to loop, for example validating user input. Make sure something
inside the loop can eventually make the condition False, or it will run forever.

```python
age = int(input("Enter your age (0 to 130): "))

while age < 0 or age > 130:
    print("That is not a valid age, try again.")
    age = int(input("Enter your age (0 to 130): "))

print(f"Thank you, you entered {age}.")
```

## break and continue

- `break` stops the loop immediately.
- `continue` skips the rest of the current pass and jumps to the next one.

```python
# break: stop as soon as we find the target
numbers = [3, 8, 1, 9, 4]
for n in numbers:
    if n == 9:
        print("Found 9, stopping")
        break
    print("checked", n)
```

```python
# continue: skip even numbers, print only odd ones
for n in range(1, 8):
    if n % 2 == 0:
        continue
    print(n)
```

## Summary

- `for item in collection:` runs the block once per item; it also works over strings.
- `range(stop)`, `range(start, stop)`, `range(start, stop, step)` generate number sequences;
  the stop value is excluded.
- `enumerate()` gives the index and the item together.
- `while condition:` repeats until the condition is False; guard against infinite loops.
- `break` stops a loop; `continue` skips to the next pass.
