## Creating strings

A **string** is text. Wrap it in single or double quotes; both work the same, so pick one
and be consistent. Use the other style when your text contains a quote.

```python
greeting = "Hello"
name = 'Sam'
sentence = "It's a sunny day"   # double quotes so the apostrophe is fine
print(greeting, name)
print(sentence)
```

## f-strings: putting values inside text

An **f-string** is a string with the letter `f` in front of the opening quote. Inside it,
anything in `{curly braces}` is replaced with the value of that variable or expression.
This is the cleanest way to build messages, and you will see it throughout the course.

```python
name = "Sam"
age = 34
print(f"{name} is {age} years old.")
print(f"Next year {name} will be {age + 1}.")
```

## Length, indexing, and slicing

`len()` gives the number of characters. Each character has a position (an **index**)
starting at `0`. Negative indexes count from the end, so `-1` is the last character.

```python
word = "Python"
print(len(word))      # 6
print(word[0])        # 'P'  (first character)
print(word[-1])       # 'n'  (last character)
```

**Slicing** takes a range of characters with `[start:end]`. The start is included and the
end is **not**. Leave a side blank to go to the beginning or the end.

```python
word = "Python"
print(word[0:3])     # 'Pyt'  positions 0, 1, 2 (3 is not included)
print(word[:3])      # 'Pyt'  same thing, start defaults to 0
print(word[3:])      # 'hon'  from position 3 to the end
print(word[-2:])     # 'on'   the last two characters
```

## Useful methods

A **method** is an action you call on a value using a dot. Strings have many. These are
the ones you will reach for most. Note that strings cannot be changed in place, so each
method returns a **new** string.

```python
text = "  Hello World  "
print(text.upper())          # '  HELLO WORLD  '
print(text.lower())          # '  hello world  '
print(text.strip())          # 'Hello World'  (removes surrounding spaces)
print(text.replace("World", "Python"))   # '  Hello Python  '
print("a,b,c".split(","))    # ['a', 'b', 'c']  (splits text into a list)
```

You can join strings with `+` and check whether one string contains another with `in`.

```python
first = "Ada"
last = "Lovelace"
full = first + " " + last
print(full)
print("Love" in full)    # True
```

## Reading input

`input()` pauses and waits for the user to type something, then returns it. The result is
**always a string**, so convert it if you need a number.

```python
name = input("What is your name? ")
print(f"Nice to meet you, {name}!")

age_text = input("How old are you? ")
age = int(age_text)            # convert the text to a number
print(f"In ten years you will be {age + 10}.")
```

## Summary

- Strings are text in single or double quotes.
- f-strings (`f"...{value}..."`) insert values into text cleanly.
- `len()` counts characters; indexing `[i]` reads one (starts at 0, `-1` is last).
- Slicing `[start:end]` takes a range; the end position is excluded.
- Methods like `.upper()`, `.lower()`, `.strip()`, `.replace()`, `.split()` return new
  strings; `+` joins and `in` tests membership.
- `input()` returns a string; convert with `int()` or `float()` for maths.
