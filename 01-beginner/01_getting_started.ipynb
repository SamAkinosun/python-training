## What is Python?

Python is a programming language: a way of writing instructions that a computer
follows. Two things make it a great first language:

- **It reads almost like English.** Less punctuation and ceremony than most languages,
  so you spend your effort on the problem, not the syntax.
- **It is everywhere.** Automating boring tasks, analysing data, building websites,
  and machine learning are all done in Python every day.

In this course you write Python inside **Jupyter notebooks**: documents that mix
explanations (like this text) with code you can run and edit.

## Running code

The grey boxes below are **code cells**. Click one and press **Shift + Enter** to run

```python
print("Hello, world!")
```

`print()` displays whatever you put between its brackets. The text in quotes is called
a **string** (more on that later). You can print numbers too, and print several things

```python
print("Learning Python")
print(2024)
print("The answer is", 42)
```

## Comments

A line that starts with `#` is a **comment**. Python ignores it. Comments are notes for
humans: use them to explain *why* the code does something.

```python
# This line is a comment and does nothing when you run the cell.
print("Comments are ignored by Python")  # you can also comment at the end of a line
```

## Reading an error

Everyone writes code that breaks. The trick is to read the error rather than panic.
Run the cell below. It refers to `mesage`, which was never created, so Python complains.

```python
message = "hi"
print(mesage)   # note the typo: 'mesage' instead of 'message'
```

Look at the **last line** of the error first. It says something like:

`NameError: name 'mesage' is not defined`

That tells you the *type* of problem (`NameError`) and the cause (the name `mesage`
does not exist). Here the fix is to correct the spelling to `message`. Reading the
bottom line of an error is the single most useful debugging habit you can build.

Fix the typo in the cell above and run it again to see it work.
