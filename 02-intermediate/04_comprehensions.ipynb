# Comprehensions

## What you'll learn
- Building a list in one readable line
- Filtering with a condition
- Building dictionaries the same way

## The idea

A very common pattern is: start with a collection, do something to each item, and collect
the results into a new list. You can write that as a loop:

```python
numbers = [1, 2, 3, 4, 5]

squares = []
for n in numbers:
    squares.append(n ** 2)
print(squares)
```

A **list comprehension** says the same thing in one line. Read it as
"the expression, for each item in the collection".

```python
numbers = [1, 2, 3, 4, 5]
squares = [n ** 2 for n in numbers]
print(squares)
```

## Adding a condition

Put an `if` at the end to keep only the items you want.

```python
numbers = range(1, 11)
evens = [n for n in numbers if n % 2 == 0]
print(evens)                       # [2, 4, 6, 8, 10]

words = ["apple", "fig", "banana", "kiwi"]
short = [w.upper() for w in words if len(w) <= 4]
print(short)                       # ['FIG', 'KIWI']
```

## Dictionary comprehensions

The same idea builds a dictionary, using `key: value` inside curly braces.

```python
numbers = [1, 2, 3, 4]
squares_map = {n: n ** 2 for n in numbers}
print(squares_map)                 # {1: 1, 2: 4, 3: 9, 4: 16}

prices = {"apple": 0.50, "banana": 0.30, "cherry": 1.20}
cheap = {name: price for name, price in prices.items() if price < 1.0}
print(cheap)                       # {'apple': 0.5, 'banana': 0.3}
```

Comprehensions are loved because they are short and clear. If one starts to get hard to
read, that is a sign to write it as a normal loop instead. Readable beats clever.

## Summary

- A list comprehension `[expr for item in collection]` builds a list in one line.
- Add `if condition` at the end to filter.
- A dictionary comprehension `{key: value for ... }` builds a dictionary the same way.
- If a comprehension gets hard to read, use a normal loop instead.
