# Files and Errors

## What you'll learn
- Reading text files safely with `with open(...)`
- Writing and appending to files
- Handling problems with `try` / `except` instead of crashing

## Reading a file

The modern, safe way to open a file is the `with` statement. It opens the file, gives it
to you as a variable, and automatically closes it when the block ends, even if something
goes wrong. A sample file, `data/Text_File.txt`, sits in this folder.

`readlines()` reads the whole file into a list, one string per line.

```python
with open("data/Text_File.txt", "r") as f:
    lines = f.readlines()

print(lines)
print(f"There are {len(lines)} lines")
```

You can also loop straight over the file to handle one line at a time. `strip()` removes
the newline character at the end of each line.

```python
with open("data/Text_File.txt", "r") as f:
    for line in f:
        print(line.strip())
```

## Writing a file

Open with mode `"w"` to **write** (this replaces any existing contents). Use `"a"` to
**append** to the end instead. Numbers must be converted to text with `str()` first.

```python
with open("output.txt", "w") as f:
    f.write("This is my output file.\n")
    f.write(str(345.65))

# Read it back to confirm
with open("output.txt", "r") as f:
    print(f.read())
```

## Handling errors

Some failures are not your fault: a file is missing, or a user types letters where a
number is expected. Wrapping risky code in `try` / `except` lets you respond calmly
instead of crashing.

```python
try:
    with open("does_not_exist.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("That file was not found. Check the name and try again.")
```

You can catch different problems separately. Here, converting the text to a number might
fail with a `ValueError`.

```python
user_text = "not a number"

try:
    number = int(user_text)
    print("You entered", number)
except ValueError:
    print(f"'{user_text}' is not a whole number.")
```

Catching the specific error type (like `FileNotFoundError` or `ValueError`) is better than
catching everything, because it only handles the problem you expected and lets genuine
bugs surface.

## Summary

- `with open(path, mode) as f:` opens a file and closes it for you when the block ends.
- Modes: `"r"` read, `"w"` write (replaces), `"a"` append. Convert numbers with `str()`
  before writing.
- `readlines()` gives a list of lines; looping over the file gives one line at a time;
  `strip()` trims the trailing newline.
- `try` / `except` handles failures gracefully; catch specific types like
  `FileNotFoundError` and `ValueError`.
