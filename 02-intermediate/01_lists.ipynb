## Creating lists

A **list** stores several items in order, in one variable. Use square brackets, with
items separated by commas. A list can hold mixed types, and you can start empty.

```python
new_customers = []                          # an empty list
customers = [2737, "Susan", 8441, "Nick"]   # mixed types are allowed
print(customers)
print(len(customers))                       # 4 items
```

## Reading items by position

Each item has an **index** starting at 0. Negative indexes count from the end.

```python
customers = [2737, "Susan", 8441, "Nick"]
print(customers[0])    # 2737 (first)
print(customers[2])    # 8441
print(customers[-1])   # 'Nick' (last)
```

## Adding and inserting

To grow a list, use `append()` to add to the end, or `insert()` to add at a position.
You cannot create a new slot with `customers[4] = ...`; that raises an `IndexError`.

```python
customers = [2737, "Susan", 8441, "Nick"]
customers.append("Mary")        # add to the end
print(customers)

customers.insert(2, "Alex")     # insert at index 2, shifting the rest along
print(customers)
```

## Removing items

`remove()` deletes the first matching value; `pop()` removes by index and returns it.

```python
customers = ["Susan", "Alex", "Nick", "Mary"]
customers.remove("Alex")     # delete by value
print(customers)

last = customers.pop()       # remove and return the last item
print(last, "->", customers)
```

## Searching, sorting, counting

- `in` checks whether a value is present.
- `sort()` orders the list in place.
- `count()` counts occurrences.
- `min()` / `max()` find the smallest / largest.

```python
customers = ["Susan", "Alex", "Nick", "Mary"]
print("Alex" in customers)      # True

customers.sort()
print(customers)                # alphabetical

letters = ["f", "a", "f", "b", "f"]
print(letters.count("f"))       # 3

numbers = [2, 4, 7, 3, 8]
print(min(numbers), max(numbers))   # 2 8
```

## Slicing

Slicing works just like it does for strings: `[start:end]` takes a range, the end is
excluded, and a blank side runs to the beginning or the end.

```python
letters = ["v", "y", "a", "j", "f", "o", "b", "w"]
print(letters[2:5])    # ['a', 'j', 'f']
print(letters[:3])     # ['v', 'y', 'a']
print(letters[-3:])    # ['o', 'b', 'w']
```

## Joining lists

`+` joins two lists into a new one.

```python
team_a = ["Susan", "Alex"]
team_b = ["Nick", "Mary"]
everyone = team_a + team_b
print(everyone)
```

## Lists inside lists

A list item can itself be a list. This is how you store rows of related data, such as
[county, town, age] for several people.

```python
donors = [
    ["Suffolk", "Ipswich", 35],
    ["Norfolk", "Thetford", 38],
    ["Essex", "Colchester", 56],
]
print(donors[1])        # ['Norfolk', 'Thetford', 38]
print(donors[1][1])     # 'Thetford' (row 1, column 1)
```

## The copy trap

This surprises everyone. Assigning one list to another with `=` does **not** make a
separate copy. Both names point to the **same** list, so changing one changes the other.

```python
a = [2, 5]
b = a            # b points to the SAME list as a, not a copy
b[0] = 9
print("a =", a)  # a = [9, 5]   <- a changed too!
print("b =", b)  # b = [9, 5]
```

To get an independent copy, take a full slice `a[:]` (or use `a.copy()`). Now the two
lists are separate and editing one leaves the other alone.

```python
a = [2, 5]
b = a[:]         # a real, independent copy
b[0] = 9
print("a =", a)  # a = [2, 5]   <- unchanged
print("b =", b)  # b = [9, 5]
```

## Summary

- Lists hold ordered, mixed items; index from 0, and `-1` is the last.
- `append`, `insert` add; `remove`, `pop` delete.
- `in`, `sort`, `count`, `min`, `max` search and organise.
- Slicing `[start:end]` excludes the end position.
- `+` joins lists; lists can be nested and read with `[row][col]`.
- `b = a` shares the same list; use `a[:]` or `a.copy()` for an independent copy.
