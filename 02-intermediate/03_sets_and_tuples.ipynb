# Sets and Tuples

## What you'll learn
- Sets: collections of unique items, and the maths you can do with them
- Tuples: fixed collections you cannot change
- When to reach for each

## Sets

A **set** is an unordered collection of **unique** items. Duplicates are dropped
automatically. Create one with curly braces.

```python
observed = {"Blackbird", "Robin", "Blackbird", "Pheasant", "Robin"}
print(observed)            # duplicates removed
print("Robin" in observed) # membership tests are very fast on sets
```

Sets support the operations you know from maths, which makes comparing two collections
easy.

```python
monday = {"Blackbird", "Robin", "Pheasant"}
tuesday = {"Starling", "Robin", "Wood Pigeon"}

print("Seen on both days:", monday & tuesday)        # intersection
print("Seen on either day:", monday | tuesday)       # union
print("Only on Monday:", monday - tuesday)           # difference
print("Seen on exactly one day:", monday ^ tuesday)  # symmetric difference
```

To start with an empty set, use `set()` (plain `{}` makes an empty dictionary, not a set).

```python
seen = set()
seen.add("Blue Tit")
seen.add("Blue Tit")    # adding a duplicate does nothing
print(seen)
```

## Tuples

A **tuple** is like a list, but it **cannot be changed** after it is created. Use round
brackets. Reach for a tuple when a group of values belongs together and should stay fixed,
such as a coordinate or a fixed record.

```python
point = (10, 20)
print(point[0])     # read by index, just like a list

# point[0] = 99     # this would raise a TypeError: tuples are read-only
```

A handy feature is **unpacking**: assigning a tuple's items to several variables at once.
You have already seen this with multiple assignment.

```python
point = (10, 20)
x, y = point        # unpack into two variables
print("x is", x, "and y is", y)
```

## Summary

- Sets hold unique, unordered items; duplicates are dropped and membership tests are fast.
- Set maths: `&` intersection, `|` union, `-` difference, `^` symmetric difference.
- Use `set()` for an empty set (`{}` is an empty dictionary).
- Tuples are fixed (read-only) lists in round brackets; great for values that belong
  together and should not change.
- Unpacking assigns a tuple's items to several variables at once.
