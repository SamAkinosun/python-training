## Why import?

You do not have to write everything yourself. Python ships with a large **standard
library** of ready-made code grouped into **modules**. You bring a module into your
notebook with `import`, then use its contents with a dot.

````python
import math

print(math.sqrt(144))   # 12.0
print(math.pi)          # 3.141592653589793
````

## Import styles

- `import math` then `math.sqrt(...)` keeps the module name as a prefix (clear about where
  things come from).
- `from math import sqrt` brings one name in directly, so you write just `sqrt(...)`.
- `import numpy as np` imports under a shorter **alias**, a common convention for big
  libraries.

````python
from math import sqrt, ceil
print(sqrt(81))     # 9.0
print(ceil(4.1))    # 5

import math as m
print(m.floor(4.9)) # 4
````

## random: pick and shuffle

The `random` module produces random choices, which is useful for sampling, games, and
test data. (Re-run the cell to see different results.)

````python
import random

print(random.randint(1, 6))                 # a dice roll: 1 to 6 inclusive
print(random.choice(["red", "green", "blue"]))   # pick one item

deck = [1, 2, 3, 4, 5]
random.shuffle(deck)                         # shuffles in place
print(deck)
````

## datetime: dates and times

The `datetime` module handles dates and times. You can get today's date, build a specific
date, and format it as text.

````python
from datetime import date

today = date.today()
print("Today is", today)
print("Year:", today.year)

independence = date(2026, 7, 4)
print("Formatted:", independence.strftime("%A, %d %B %Y"))
````
