## Why functions?

When the same steps appear in several places, copying them is risky: if the logic needs to
change, you must find and edit every copy, and it is easy to miss one. A **function** lets
you write the steps once, give them a name, and reuse them. This is the **DRY** idea: Do
not Repeat Yourself.

Below, the same number check is repeated three times. That repetition is the problem a
function solves.

```python
# Repetitive version (the problem)
results = []

value = 10
if value > 0:
    results.append(value)

value = -3
if value > 0:
    results.append(value)

value = 7
if value > 0:
    results.append(value)

print(results)
```

## Defining a function

Use `def`, a name, parentheses for any inputs (**parameters**), and a colon. The indented
body runs when you **call** the function. `return` sends a value back to the caller.

```python
def is_positive(number):
    return number > 0

print(is_positive(10))    # True
print(is_positive(-3))    # False
```

Now the repetitive code becomes short and there is only one place to change the rule.

```python
def is_positive(number):
    return number > 0

results = [value for value in [10, -3, 7] if is_positive(value)]
print(results)            # [10, 7]
```

## Parameters and return values

A function can take several parameters and return a computed result.

```python
def rectangle_area(width, height):
    return width * height

area = rectangle_area(5, 3)
print(area)               # 15
```

## Default values and keyword arguments

A parameter can have a **default**, used when the caller does not supply it. You can also
pass arguments **by name**, which is clearer when there are several.

```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Maya"))                     # Hello, Maya!
print(greet("Maya", "Welcome"))          # Welcome, Maya!
print(greet(greeting="Hi", name="Ada"))  # Hi, Ada!  (named, so order does not matter)
```

## Accepting any number of arguments

`*args` collects extra positional arguments into a tuple, and `**kwargs` collects extra
named ones into a dictionary. You will see these in library code; here is the idea.

```python
def total(*numbers):
    return sum(numbers)

print(total(1, 2, 3))         # 6
print(total(10, 20, 30, 40))  # 100
```

## Variable scope

A variable created **inside** a function exists only inside it; this is its **scope**.
Trying to use it outside raises a `NameError`. Pass data in through parameters and get
results out with `return`, rather than relying on variables leaking across that boundary.

```python
def make_message():
    secret = "inside only"
    return secret

print(make_message())     # works: the value is returned
# print(secret)           # would raise NameError: 'secret' lives only inside the function
```

## Docstrings

A string just under the `def` line documents what the function does. Tools and editors
show it as help, and `help(function)` prints it.

```python
def celsius_to_fahrenheit(celsius):
    """Convert a temperature in Celsius to Fahrenheit."""
    return celsius * 9 / 5 + 32

print(celsius_to_fahrenheit(20))   # 68.0
help(celsius_to_fahrenheit)
```

## Summary

- Functions let you write logic once and reuse it (the DRY idea).
- `def name(parameters):` defines one; `return` sends a value back.
- Parameters can have defaults, and arguments can be passed by name.
- `*args` / `**kwargs` accept any number of positional / named arguments.
- Variables made inside a function are local to it (scope); use parameters and `return`
  to move data in and out.
- A docstring under `def` documents the function.

## Practice Solutions

```python
# 1.
def double(n):
    return n * 2
print(double(4), double(10))

# 2.
def full_name(first, last):
    return first + " " + last
print(full_name("Ada", "Lovelace"))

# 3.
def greet(name, greeting="Hi"):
    return f"{greeting}, {name}!"
print(greet("Maya"))
print(greet("Maya", "Welcome"))

# 4.
def average(*numbers):
    return sum(numbers) / len(numbers)
print(average(2, 4, 9))

# 5.
def to_initials(name):
    """Return the upper-case initials of each word in name."""
    return "".join(word[0].upper() for word in name.split())
print(to_initials("ada lovelace"))   # AL
```
