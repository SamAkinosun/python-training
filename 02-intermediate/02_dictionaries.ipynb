# Dictionaries

## What you'll learn
- Storing data as key/value pairs
- Looking up, adding, updating, and removing values
- Safe lookups with `get()`
- Looping over keys, values, and pairs

## What a dictionary is

A **dictionary** stores **key/value pairs**. Instead of looking items up by position like
a list, you look them up by a meaningful key. Use curly braces with `key: value` pairs.

```python
blackbird = {
    "genus": "Turdus",
    "species": "merula",
    "length_cm": 24,
}
print(blackbird)
print(blackbird["species"])    # look up by key -> 'merula'
```

## Adding and updating

Assigning to a key updates it if it exists, or creates it if it does not.

```python
prices = {"apple": 0.50, "banana": 0.30}
prices["cherry"] = 1.20      # add a new pair
prices["apple"] = 0.55       # update an existing one
print(prices)
```

## Removing

`del` removes a pair by key; `pop()` removes it and gives you the value back.

```python
prices = {"apple": 0.55, "banana": 0.30, "cherry": 1.20}
del prices["banana"]
print(prices)

removed = prices.pop("cherry")
print("removed", removed, "->", prices)
```

## Safe lookups with get()

Looking up a missing key with square brackets raises a `KeyError`. `get()` returns
`None` (or a default you choose) instead, which avoids crashes.

```python
prices = {"apple": 0.55}
print(prices.get("banana"))          # None, no error
print(prices.get("banana", 0.0))     # 0.0, your chosen default
print("apple" in prices)             # True: 'in' checks the keys
```

## Looping

By default, looping over a dictionary gives you its **keys**. Use `.values()` for the
values and `.items()` for both at once.

```python
bird = {"genus": "Turdus", "species": "merula", "length_cm": 24}

for key in bird:
    print(key, "->", bird[key])

print("---")

for key, value in bird.items():
    print(f"{key}: {value}")
```

## Summary

- Dictionaries store `key: value` pairs in curly braces; look up by key, not position.
- Assigning to a key adds or updates it; `del` and `pop()` remove.
- `get(key, default)` avoids `KeyError` on missing keys; `in` tests the keys.
- Loop keys directly, `.values()` for values, `.items()` for key and value together.
