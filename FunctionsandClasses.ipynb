# Functions

Functions can be useful in several different scenarios:
* **When you have sections of code which are repeated more than once.**
    * This is inefficient. If you need to make a change to the repeated code you have to go through all your code and find where those repeated block are and change each one individually. A prime opportunity for mistakes to be made.
    * you can place the repeated code into a function and give the function a name. Now you cal call the function wherever you ned that code. This means the repeated code is in one place only and much easier to edit/manage.
* Make Break complicated section of code down into smaller manageable sections
* Code placed in a function is easier to share with other teams/projects.

When you have code with repeated sections of code, this is called **IMPERATIVE PROGRAMMING"**. It can be easier to read but introduces inefficiencies with repeated blocks of code. The code below shows an example of an imperative style of coding, where each request for input is coded individually.

```python
####################################################################
# IMPERATIVE Example
####################################################################

my_list = []  # Create empty list

# Ask for first number
my_input = int(input("Enter a number: "))

# Check the input number
if my_input > 0:
    print(f"{my_input} is a positive number, added to list")
    my_list.append(my_input)
elif my_input == 0:
    print("You entered zero, cannot add it to list")
else:
    print(f"{my_input} is a negative number, cannot add it to list")

# ask for second number and check its value
my_input = int(input("Enter a number: "))
if my_input > 0:
    print(f"{my_input} is a positive number, added to list")
    my_list.append(my_input)
elif my_input == 0:
    print("You entered zero, cannot add it to list")
else:
    print(f"{my_input} is a negative number, cannot add it to list")

# ask for third number and check its value
my_input = int(input("Enter a number: "))
if my_input > 0:
    print(f"{my_input} is a positive number, added to list")
    my_list.append(my_input)
elif my_input == 0:
    print("You entered zero, cannot add it to list")
else:
    print(f"{my_input} is a negative number, cannot add it to list")

# Now print the whole list
print("Your List:",my_list)
```

To improve on the above approach, you can place the repeated code into a function, thereby taking a **FUNCTIONAL PROGRAMMING** approach.

```python
####################################################################
# FUNCTIONAL Example
####################################################################

def my_check(in_num):
    if in_num > 0:
        print(f"{in_num} is a positive number, added to list")
        return True
    elif in_num == 0:
        print("You entered zero, cannot add it to list")
        return False
    else:
        print(f"{in_num} is a negative number, cannot add it to list")
        return False  # function returns a single boolean value


my_list = []  # Create empty list
counter = 0

while counter < 3:
    # Ask for first number
    my_input = int(input("Enter a number: "))

    result = my_check(my_input)
    if result:
        my_list.append(my_input)
        counter = counter + 1

# Now print the whole list
print("Your List:",my_list)
```

# CLASSES

Classes are a way of modeling real world objects or concepts. A class is similar to a blueprint for a building or vehicle. you can use the class as a set of instructions on how to create many copies of the same thing. A class can have **attributes (variables)** which model characteristics (height weight etc) and **behaviours (methods)** which model what that class can actually do. the example slides below demonstrate how a class can be constructed for a vehicle.

The process of taking a class and create and object from it is called **instantiation**.
* CLASS = set of instructions
* OBJECT = an item created (instatiated) from the class (instructions)
* METHOD = similar to a function but exists inside a class

A simple example is shown below:

```python
class A:
    def foo(self):
        print('Foo')

    def bar(self, my_argument):
        print(my_argument)

a = A()
a.foo()
a.bar('Boo') 
```

An extended example:

```python
# Define the Vehicle Class
class Vehicle:
    # A simple attempt to model a dog
    def __init__(self, name, length):
        # Initialize name and age attributes
        self.name = name
        self.length = length

    def move_forward(self):
        # Simulate vehicle moving forward
        print(f"{self.name} is now moving forward.")

    def stop(self):
        # Simulate vehicle stopping
        print(f"{self.name} has stopped")

    def crash(self):
        print(f"{self.name} has crashed")
        self.stop()

# INSTANTIATE a new OBJECT from the CLASS
my_vehicle = Vehicle("Ducati", 2.1)
my_other_vehicle = Vehicle("Mercedes", 4.6)

# Extract data from the OBJECT
print(f"My main vehicle is a {my_vehicle.name}")
print(f"My second vehicle is a {my_other_vehicle.name}")

# Move my main vehicle forward
my_vehicle.move_forward()
```

## Inheritance Example

```python
# Define the Vehicle Class
class Vehicle:

    # Initialize variables
    def __init__(self, name, length):
        self.name = name
        self.length = length

    # Simulate vehicle moving forward
    def move_forward(self):
        print(f"{self.name} is now moving forward.")

    # Simulate vehicle stopping
    def stop(self):
        print(f"{self.name} has stopped")

    # Simulate a crash
    def crash(self):
        print(f"{self.name} has crashed")
        self.stop()


# Define motorbike class
class MotorBike(Vehicle):

    # Initialize variables
    def __init__(self, name, length, colour):
        super().__init__(name, length)   # Call init() from parent class
        self.colour = colour             # define new variable which is not available in parent class

    def wheelie(self):
        super().move_forward()          # Call parent method, uses the name variable of this class, not the parents
        print(f"{self.name} is doing a wheelie")


# ######### Main code

# INSTANTIATE a new OBJECT from the CLASS
my_vehicle = Vehicle("Ducati", 2.1)
my_other_vehicle = Vehicle("Mercedes", 4.6)
my_bike = MotorBike("Yamaha", 190, "Red")

# Extract data from the OBJECT
print(f"My main vehicle is a {my_vehicle.name}")
print(f"My second vehicle is a {my_other_vehicle.name}")

# Move my main vehicle forward
my_vehicle.move_forward()

# Stop my main vehicle
my_vehicle.crash()

# Make motorbike do a wheelie
my_bike.wheelie()
```

You must be careful with which variables you expect to be able to access within a class, called **variable scope**. variables defined within a method are only accessible from within that method.<br>
Variables defined outside of a method are accessible from within a method.

## Overriding a Parent Method

When a class inherits methods from a parent class, you can give the class a customised version of an inherited method. This is called **overriding** the method. To do this you simply create a new definition for that method. The code from the previous example has been modified below to override the `move_forward()` method.

```python
# Define the Vehicle Class
class Vehicle:

    # Initialize variables
    def __init__(self, name, length):
        self.name = name
        self.length = length

    # Simulate vehicle moving forward
    def move_forward(self):
        print(f"{self.name} is now moving forward.")

    # Simulate vehicle stopping
    def stop(self):
        print(f"{self.name} has stopped")

    # Simulate a crash
    def crash(self):
        print(f"{self.name} has crashed")
        self.stop()


# Define motorbike class
class MotorBike(Vehicle):

    # Initialize variables
    def __init__(self, name, length, colour):
        super().__init__(name, length)   # Call init() from parent class
        self.colour = colour             # define new variable which is not available in parent class

    def wheelie(self):
        # ####################################
        # Call the overiden move_forward method in this class, NOT THE PARENTS
        # ####################################       
        self.move_forward()           
        print(f"{self.name} is doing a wheelie")
    
    # ####################################
    # Overriding the parent method
    # ####################################
    def move_forward(self):
        print(f"The {self.colour} {self.name} is now moving forward.")


# ######### Main code

# INSTANTIATE a new OBJECT from the CLASS
my_vehicle = Vehicle("Ducati", 2.1)
my_other_vehicle = Vehicle("Mercedes", 4.6)
my_bike = MotorBike("Yamaha", 190, "Red")

# Extract data from the OBJECT
print(f"My main vehicle is a {my_vehicle.name}")
print(f"My second vehicle is a {my_other_vehicle.name}")

# Move my main vehicle forward
my_vehicle.move_forward()

# Stop my main vehicle
my_vehicle.crash()

# Make motorbike do a wheelie
my_bike.wheelie()
```

# FILES

Being able to read and write to text files will be an important ability. This is fairly straight forward in Python.

```python
input_file = open("Text_File.txt", "r")
lines = input_file.readlines()
print(lines)
input_file.close()
```

Note how the input from the text file is displayed. Look carefully, and you will see that the square brackets and items separated by a comma show that the lines of the text file have been saved into a list. We can check this:

```python
input_file = open("Text_File.txt", "r")
lines = input_file.readlines()
print(type(lines))   # Check the datat type of the input
input_file.close()
```

To discover how many lines are in a file:

```python
input_file = open("Text_File.txt", "r")
lines = input_file.readlines()
print(f"\nThere are {len(lines)} lines in the text file")
input_file.close()
```

As the input is saved into a list, we already know how to loop through each item in a list. Note the different way in which the input is displayed using this method, compared the previous examples.

```python
input_file = open("Text_File.txt", "r")
lines = input_file.readlines()
for l in lines:   # Loop through the list, print each line as we go
    print(l)
input_file.close()
```

The input is a list, so we can access individual lines:

```python
input_file = open("Text_File.txt", "r")
lines = input_file.readlines()
print(f"\nThe second line is: {lines[1]}") # print just line 2
input_file.close()
```

Writing data out to a text file is also straight forward using the  `write()` method.

```python
output_file = open("Output.txt", "w")
my_string = "This is my output file.\n"
my_number = 345.65
output_file.write(my_string)
output_file.write(str(my_number))
output_file.close()
```

To append or insert data into a text file, you change the `w` parameter according to your needs, see the documentation page for more details.: https://docs.python.org/3/library/functions.html?highlight=open#open
