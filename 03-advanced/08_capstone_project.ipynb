## Worked solution

One way to do the whole thing. Yours may look different and still be correct.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: load and inspect
df = pd.read_csv("data/populations.csv").set_index("County")
print("Counties:", df.shape[0])
print("Columns:", df.columns.tolist())
print(df.head())

# Step 2: largest in 2019
print("\nLargest in 2019:", df["2019"].idxmax(), "->", df["2019"].max())

# Step 3: growth, absolute and percentage
df["growth"] = df["2019"] - df["1831"]
df["growth_pct"] = (df["growth"] / df["1831"] * 100).round(1)
print("\n", df)

# Step 4: fastest growing by percentage
fastest = df["growth_pct"].idxmax()
print("\nFastest growing by percent:", fastest, "->", df.loc[fastest, "growth_pct"], "%")

# Step 5: chart
ax = df[["1831", "2019"]].plot(kind="bar", title="County population: 1831 vs 2019")
ax.set_xlabel("County")
ax.set_ylabel("Population")
plt.tight_layout()
plt.show()
```
