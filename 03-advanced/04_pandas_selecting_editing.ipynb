# Pandas: Selecting and Editing

## What you'll learn
- Selecting columns, rows, and single cells with `loc`
- Filtering rows with a condition
- Adding and removing columns and rows
- Setting a meaningful index

We will use the populations table throughout. Run this first so `df` exists.

```python
import pandas as pd

df = pd.read_csv("data/populations.csv")
print(df)
```

## Selecting a column

Square brackets with a column name return that column as a Series.

```python
print(df["1831"])
```

## Setting a meaningful index

By default rows are numbered `0, 1, 2, ...`. Setting a column as the index lets you look up
rows by a real label. Here we use the county name.

```python
df = df.set_index("County")
print(df)
```

## Selecting rows and cells with loc

`loc` selects by label. The pattern is `df.loc[rows, columns]`. A colon `:` means "all".

```python
print(df.loc["Norfolk", :])        # one whole row
print("---")
print(df.loc["Norfolk", "1831"])   # one cell
print("---")
print(df.loc[:, "2019"])           # one whole column
```

## Filtering rows with a condition

Put a condition inside `loc` to keep only the rows that match. This is the everyday way to
ask questions of your data.

```python
big = df.loc[df["2019"] > 860000, :]
print(big)
```

## Adding a column

Assigning to a new column name creates it. Here we compute growth from 1831 to 2019.

```python
df["growth"] = df["2019"] - df["1831"]
print(df)
```

## Adding a row

In current pandas, build the new row as its own small DataFrame and join it with
`pd.concat`. (The old `df.append(...)` method was removed in pandas 2.0.)

```python
import pandas as pd

new_row = pd.DataFrame(
    {"1831": [600654], "2019": [900343], "growth": [299689]},
    index=["Hertfordshire"],
)
df = pd.concat([df, new_row])
print(df)
```

## Removing a column or row

`drop` removes by label. Use `columns=` to drop a column and `index=` to drop a row.

```python
df = df.drop(columns="growth")     # remove the growth column
df = df.drop(index="Hertfordshire")  # remove the row we added
print(df)
```

## Summary

- `df["col"]` selects a column.
- `df.set_index("col")` makes a column the row labels.
- `df.loc[rows, cols]` selects by label; `:` means all; a condition filters rows.
- Assigning to a new name adds a column; `pd.concat` adds rows (not `df.append`).
- `df.drop(columns=...)` and `df.drop(index=...)` remove columns and rows.
