## Why NumPy?

NumPy is the foundation of data work in Python. Its core object is the **array**: a
container for numbers that is far faster and more convenient than a list when you are doing
maths. pandas (the next lessons) is built on top of it.

The headline feature is **vectorised** operations: you apply a calculation to every element
at once, with no loop.

```python
import numpy as np

# A loop on a list multiplies one item at a time
prices = [10, 20, 30]
print([p * 1.2 for p in prices])

# A NumPy array applies the maths to everything at once
arr = np.array([10, 20, 30])
print(arr * 1.2)
```

## Creating arrays

Make an array from a list, or generate one with helpers like `arange` (a range) and
`linspace` (evenly spaced points).

```python
import numpy as np

print(np.array([1, 2, 3, 4]))
print(np.zeros(5))            # five zeros
print(np.arange(0, 10, 2))   # 0, 2, 4, 6, 8
print(np.linspace(0, 1, 5))  # 5 evenly spaced points from 0 to 1
```

## Maths on whole arrays

Arithmetic and comparisons apply element by element. Two arrays of the same size combine
position by position.

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print(a + b)        # [11 22 33 44]
print(a ** 2)       # [ 1  4  9 16]
print(a > 2)        # [False False  True  True]
```

## Indexing and slicing

Indexing and slicing look like lists. You can also select with a condition (**boolean
indexing**), which is how you filter data.

```python
import numpy as np

arr = np.array([5, 10, 15, 20, 25])
print(arr[0])         # 5
print(arr[1:4])       # [10 15 20]
print(arr[arr > 12])  # [15 20 25]  keep only values above 12
```

## Summarising

Arrays have built-in summaries that are fast and read clearly.

```python
import numpy as np

arr = np.array([4, 8, 15, 16, 23, 42])
print("sum:", arr.sum())
print("mean:", arr.mean())
print("min/max:", arr.min(), arr.max())
print("standard deviation:", round(arr.std(), 2))
```

## Summary

- A NumPy array holds numbers and supports fast, loop-free (vectorised) maths.
- Create with `np.array`, `np.zeros`, `np.arange`, `np.linspace`.
- Arithmetic and comparisons apply to every element; same-size arrays combine elementwise.
- Index and slice like lists; filter with a condition (`arr[arr > 12]`).
- `.sum()`, `.mean()`, `.min()`, `.max()`, `.std()` summarise an array.
