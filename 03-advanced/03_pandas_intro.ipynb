# Pandas: Series and DataFrames

## What you'll learn
- The two core pandas objects: `Series` and `DataFrame`
- Building them from lists, dictionaries, and CSV files

## What pandas is for

pandas is the standard tool for working with tables of data in Python: loading,
cleaning, filtering, summarising. It has two core objects:

- a **Series**: a single column of data with labels
- a **DataFrame**: a whole table (think of a spreadsheet)

```python
import pandas as pd

data = pd.Series([0.25, 0.5, 0.75, 1.0])
print(data)
```

A Series has an **index** (the labels on the left). By default it is `0, 1, 2, ...`, but
you can supply your own, which makes lookups read like a dictionary.

```python
import pandas as pd

data = pd.Series([0.25, 0.5, 0.75, 1.0], index=["a", "b", "c", "d"])
print(data)
print("Value at 'c':", data["c"])
```

## DataFrames from dictionaries

A common way to build a table is from a dictionary, where each key becomes a column name.

```python
import pandas as pd

df = pd.DataFrame({
    "song": ["In Flagranti", "Baris K", "Cemalim", "One Night"],
    "loudness": [-7.37, -8.62, -11.06, -12.69],
})
print(df)
```

You can also build a DataFrame from several dictionaries that share the same keys. The
shared keys become the row labels.

```python
import pandas as pd

pop_1831 = {"Dallas": 536853, "Minneapolis": 979000, "Chicago": 1292300}
pop_2019 = {"Dallas": 855800, "Minneapolis": 1846660, "Chicago": 907760}

df = pd.DataFrame({"1831": pop_1831, "2019": pop_2019})
print(df)
```

## Loading a CSV file

Most of the time your data lives in a file. `read_csv` loads a CSV straight into a
DataFrame. The file `data/populations.csv` sits in this folder.

```python
import pandas as pd

df = pd.read_csv("data/populations.csv")
print(df)
```
