## The idea

A **class** is a blueprint. It describes what data a thing holds (**attributes**) and what
it can do (**methods**). From one class you can create many **objects**, each with its own
data. Think of a class as the design for a vehicle, and each object as a specific vehicle
built from it.

- **class** = the blueprint
- **object** = a thing built from the blueprint
- **method** = a function that belongs to the class

## A first class

`__init__` is a special method that runs when you create an object. It sets up the
object's attributes. `self` refers to the particular object being worked on.

```python
class Vehicle:
    def __init__(self, name, length):
        # Set up this vehicle's data
        self.name = name
        self.length = length

    def move_forward(self):
        print(f"{self.name} is now moving forward.")

    def stop(self):
        print(f"{self.name} has stopped.")


# Create (instantiate) two objects from the class
car = Vehicle("Mercedes", 4.6)
bike = Vehicle("Ducati", 2.1)

print(car.name)        # read an attribute
print(bike.length)
car.move_forward()     # call a method
bike.stop()
```

Each object keeps its own attribute values, which is why `car.name` and `bike.name`
differ even though both come from the same class.

## Inheritance

A class can build on another. A **child** class inherits the parent's attributes and
methods, and can add its own. `super().__init__(...)` runs the parent's setup so you do
not repeat it.

```python
class Vehicle:
    def __init__(self, name, length):
        self.name = name
        self.length = length

    def move_forward(self):
        print(f"{self.name} is now moving forward.")


class MotorBike(Vehicle):
    def __init__(self, name, length, colour):
        super().__init__(name, length)   # run the parent's setup
        self.colour = colour             # add something new

    def wheelie(self):
        self.move_forward()              # reuse an inherited method
        print(f"The {self.colour} {self.name} is doing a wheelie!")


bike = MotorBike("Yamaha", 1.9, "red")
bike.wheelie()
```

## Overriding

A child class can replace an inherited method by defining one with the same name. This is
**overriding**. Here `MotorBike` gives its own version of `move_forward`.

```python
class Vehicle:
    def __init__(self, name):
        self.name = name

    def move_forward(self):
        print(f"{self.name} is now moving forward.")


class MotorBike(Vehicle):
    def __init__(self, name, colour):
        super().__init__(name)
        self.colour = colour

    # Same name as the parent's method -> overrides it
    def move_forward(self):
        print(f"The {self.colour} {self.name} roars forward!")


Vehicle("Truck").move_forward()          # parent version
MotorBike("Yamaha", "red").move_forward() # overridden version
```

## Summary

- A class is a blueprint; objects are built from it, each with its own attribute values.
- `__init__(self, ...)` sets up a new object; `self` is the object itself.
- Methods are functions defined inside the class and called with a dot.
- A child class inherits from a parent; `super().__init__(...)` runs the parent's setup.
- Defining a method with the same name as an inherited one overrides it.

```python
# 1 and 2.
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        print(f"{self.name} says woof!")

    def describe(self):
        print(f"{self.name} is a {self.breed}")

a = Dog("Rex", "Labrador")
b = Dog("Bella", "Poodle")
a.bark()
b.describe()

# 3.
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

print(Circle(5).area())

# 4.
class Puppy(Dog):
    def bark(self):
        print(f"{self.name} says yip!")

Puppy("Tiny", "Chihuahua").bark()
```
