# Visualising Data

## What you'll learn
- Drawing quick charts straight from a DataFrame
- Bar charts, line charts, and comparisons
- Labelling a chart so it is readable

Charts make patterns obvious that are hard to see in a table. pandas can draw them
directly using matplotlib underneath. In JupyterLab the charts appear right under the cell.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/populations.csv").set_index("County")
print(df)
```

## A bar chart

A bar chart is the natural choice for comparing a value across categories. Here we compare
the 2019 population of each county.

```python
ax = df["2019"].plot(kind="bar", title="Population by county (2019)")
ax.set_xlabel("County")
ax.set_ylabel("Population")
plt.tight_layout()
plt.show()
```

## Comparing two columns

Plotting the whole DataFrame draws one set of bars per column, so you can compare 1831 with
2019 side by side. This is the right chart because we are comparing the same counties across
two points in time.

```python
ax = df.plot(kind="bar", title="County population: 1831 vs 2019")
ax.set_xlabel("County")
ax.set_ylabel("Population")
plt.tight_layout()
plt.show()
```

## A horizontal bar chart of a computed value

You can chart a column you compute on the fly. Growth is easier to read as a horizontal bar
chart when the category labels are long.

```python
df["growth"] = df["2019"] - df["1831"]
ax = df["growth"].sort_values().plot(kind="barh", title="Population growth 1831 to 2019")
ax.set_xlabel("Increase in population")
plt.tight_layout()
plt.show()
```

## A line chart

Line charts suit a trend over an ordered sequence. With only two time points the line is
short, but the idea scales to many points. We transpose the table so the years run along
the bottom and each county is a line.

```python
ax = df[["1831", "2019"]].T.plot(kind="line", marker="o", title="Population trend by county")
ax.set_xlabel("Year")
ax.set_ylabel("Population")
plt.tight_layout()
plt.show()
```

## Choosing a chart

- **Bar**: compare a value across categories.
- **Grouped bar**: compare a few series across the same categories.
- **Line**: show a trend over an ordered sequence (often time).
- **Horizontal bar**: like a bar chart, but easier when labels are long or there are many.

Always give a chart a title and axis labels so it stands on its own.

## Summary

- `series.plot(kind="bar")` and `df.plot(kind="bar")` draw charts straight from pandas.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/populations.csv").set_index("County")

# 1.
df["1831"].plot(kind="bar", title="Population by county (1831)")
plt.tight_layout()
plt.show()

# 2.
ax = df.plot(kind="bar", title="1831 vs 2019")
ax.set_xlabel("County")
ax.set_ylabel("Population")
plt.tight_layout()
plt.show()

# 3.
df["growth"] = df["2019"] - df["1831"]
df["growth"].sort_values().plot(kind="barh", title="Growth")
plt.tight_layout()
plt.show()

# 4.
sales = pd.DataFrame({
    "region": ["North", "South", "North", "East", "South"],
    "revenue": [100, 75, 120, 130, 70],
})
sales.groupby("region")["revenue"].sum().plot(kind="bar", title="Revenue by region")
plt.tight_layout()
plt.show()
```
