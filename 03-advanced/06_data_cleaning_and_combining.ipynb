# Data Cleaning and Combining

## What you'll learn
- Finding and filling missing values
- Fixing column types
- Renaming columns
- Combining two tables with `merge`

Real data is messy. Before any analysis you usually spend time cleaning it. We build a
small messy table to practise on.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Ann", "Ben", "Cara", "Dan"],
    "age": [30, np.nan, 41, 25],          # one missing age
    "city": ["Oslo", "Lima", None, "Rome"],  # one missing city
    "spend": ["100", "250", "75", "300"],  # numbers stored as text!
})
print(df)
```

## Finding missing values

`isna()` flags missing cells; summing it per column counts them.

```python
print(df.isna())
print("---")
print(df.isna().sum())   # missing count per column
```

## Filling or dropping missing values

You can drop rows with missing data, or fill the gaps. Filling is often better so you do
not throw away whole rows. Fill the numeric gap with the average age, and the text gap with
a placeholder.

```python
df["age"] = df["age"].fillna(df["age"].mean())
df["city"] = df["city"].fillna("Unknown")
print(df)
```

## Fixing column types

The `spend` column holds numbers stored as text, so maths would misbehave. Convert it to a
real numeric type with `astype`.

```python
print("before:", df["spend"].dtype)
df["spend"] = df["spend"].astype(int)
print("after:", df["spend"].dtype)
print("total spend:", df["spend"].sum())
```

## Renaming columns

Clear column names make everything downstream easier. `rename` changes the ones you name.

```python
df = df.rename(columns={"spend": "spend_usd", "name": "customer"})
print(df.columns.tolist())
```

## Combining two tables

Often the information you need is split across tables. `merge` joins them on a shared
column, like a lookup. Here we attach each city's country.

```python
import pandas as pd

people = pd.DataFrame({
    "name": ["Ann", "Ben", "Cara"],
    "city": ["Oslo", "Lima", "Rome"],
})
countries = pd.DataFrame({
    "city": ["Oslo", "Lima", "Rome"],
    "country": ["Norway", "Peru", "Italy"],
})

combined = people.merge(countries, on="city")
print(combined)
```

## Summary

- `isna()` flags missing cells; `.isna().sum()` counts them per column.
- `fillna(value)` fills gaps; `dropna()` removes rows with gaps.
- `astype(type)` converts a column's type (for example text to int).
- `rename(columns={...})` renames columns.
- `left.merge(right, on="col")` joins two tables on a shared column.

```python
import pandas as pd
import numpy as np

# 1.
df = pd.DataFrame({"x": [1, np.nan, 3], "y": ["a", "b", None]})
print(df.isna().sum())

# 2.
df["x"] = df["x"].fillna(0)
df["y"] = df["y"].fillna("missing")
print(df)

# 3.
df["count"] = ["10", "20", "30"]
df["count"] = df["count"].astype(int)
print(df["count"].sum())   # 60

# 4.
df = df.rename(columns={"x": "value"})
print(df.columns.tolist())

# 5.
left = pd.DataFrame({"id": [1, 2], "name": ["Ann", "Ben"]})
right = pd.DataFrame({"id": [1, 2], "score": [90, 80]})
print(left.merge(right, on="id"))
```
