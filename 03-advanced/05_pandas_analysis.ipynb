# Pandas: Exploring and Summarising

## What you'll learn
- The first commands to run on any new dataset
- Quick statistics with `describe`
- Counting categories with `value_counts`
- Grouping and aggregating with `groupby`
- Sorting results

For this lesson we use a slightly larger sample so grouping is meaningful: a small table of
sales, built inline.

```python
import pandas as pd

sales = pd.DataFrame({
    "region": ["North", "South", "North", "East", "South", "North", "East"],
    "product": ["A", "B", "B", "A", "A", "A", "B"],
    "units": [10, 5, 8, 12, 7, 3, 9],
    "revenue": [100, 75, 120, 130, 70, 30, 140],
})
print(sales)
```

## First look at a dataset

When you open a new dataset, these four are the first things to run.

```python
print(sales.head(3))     # the first few rows
print("---")
print(sales.shape)       # (rows, columns)
print("---")
print(sales.columns.tolist())   # the column names
```

```python
sales.info()             # column names, types, and non-missing counts
```

## Quick statistics

`describe()` reports count, mean, min, max, and quartiles for every numeric column at once.

```python
print(sales.describe())
```

## Counting categories

`value_counts()` tallies how often each value appears in a column, which is the fastest way
to understand a category.

```python
print(sales["region"].value_counts())
```

## Grouping

`groupby` splits the data into groups, applies a summary to each, and combines the results.
"This per that" questions are almost always a groupby: total revenue per region, average
units per product, and so on.

```python
# Total revenue per region
print(sales.groupby("region")["revenue"].sum())
print("---")
# Average units per product
print(sales.groupby("product")["units"].mean())
```

You can group by more than one column and summarise more than one column at once.

```python
print(sales.groupby(["region", "product"])["revenue"].sum())
```

## Sorting

`sort_values` orders rows by a column. Add `ascending=False` for largest first.

```python
print(sales.sort_values("revenue", ascending=False))
```

## Summary

- Start a new dataset with `head()`, `shape`, `columns`, and `info()`.
- `describe()` gives summary statistics for all numeric columns.
- `value_counts()` tallies a category column.
- `groupby("col")["other"].agg()` answers "this per that" questions (`sum`, `mean`, etc.).
- `sort_values("col", ascending=False)` orders the rows.
