# Python Training - Start Here

Welcome. This is a self-paced introduction to Python for people who are new to
programming. You learn by reading short explanations, running real code, and doing
practice exercises, all inside these notebooks.

**How to use a notebook:** click a code cell and press **Shift + Enter** to run it. Change
the code and run it again. Each lesson ends with practice exercises and their solutions.

**Setup (once):** from a terminal in this folder, run `pip install -r requirements.txt`
(or `conda install jupyterlab numpy pandas matplotlib`), then `jupyter lab`.

Work through the three stages in order. The links below open each lesson.

## Stage 1: Beginner

The foundations. No prior experience needed.

1. [Getting Started](01-beginner/01_getting_started.ipynb) - run code, print output, comments, reading errors
2. [Variables and Types](01-beginner/02_variables_and_types.ipynb) - `int`, `float`, `str`, `bool`, conversion
3. [Numbers and Math](01-beginner/03_numbers_and_math.ipynb) - operators and order of operations
4. [Strings](01-beginner/04_strings.ipynb) - f-strings, slicing, methods, `input()`
5. [Making Decisions](01-beginner/05_making_decisions.ipynb) - comparisons and `if` / `elif` / `else`
6. [Loops](01-beginner/06_loops.ipynb) - `for`, `while`, `break`, `continue`

## Stage 2: Intermediate

Collections, code organisation, and files.

1. [Lists](02-intermediate/01_lists.ipynb) - build, edit, slice, nest, and the copy trap
2. [Dictionaries](02-intermediate/02_dictionaries.ipynb) - key/value pairs
3. [Sets and Tuples](02-intermediate/03_sets_and_tuples.ipynb) - unique items and fixed groups
4. [Comprehensions](02-intermediate/04_comprehensions.ipynb) - build lists and dicts in one line
5. [Functions](02-intermediate/05_functions.ipynb) - parameters, defaults, return, scope
6. [Modules and Imports](02-intermediate/06_modules_and_imports.ipynb) - reuse standard-library code
7. [Files and Errors](02-intermediate/07_files_and_errors.ipynb) - read/write files, `try` / `except`

## Stage 3: Advanced

Object-oriented programming and the data stack (NumPy, pandas, matplotlib). Make sure you
have run `pip install -r requirements.txt` first.

1. [Classes](03-advanced/01_classes.ipynb) - attributes, methods, inheritance
2. [NumPy](03-advanced/02_numpy.ipynb) - fast numeric arrays
3. [Pandas: Series and DataFrames](03-advanced/03_pandas_intro.ipynb) - load and build tables
4. [Pandas: Selecting and Editing](03-advanced/04_pandas_selecting_editing.ipynb) - filter, add, remove
5. [Pandas: Exploring and Summarising](03-advanced/05_pandas_analysis.ipynb) - `describe`, `value_counts`, `groupby`
6. [Data Cleaning and Combining](03-advanced/06_data_cleaning_and_combining.ipynb) - missing values, types, `merge`
7. [Visualising Data](03-advanced/07_visualizing_data.ipynb) - bar, line, and comparison charts
8. [Capstone Project](03-advanced/08_capstone_project.ipynb) - a short end-to-end analysis

## Tips

- Run every cell as you read; do not just skim. Breaking code on purpose teaches a lot.
- Do the practice before looking at the solutions at the bottom of each notebook.
- If a cell errors, read the **last line** of the message first; it names the problem.
- Each stage folder has its own README with the same lesson list and objectives.

Good luck, and have fun.
