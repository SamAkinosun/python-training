## The workflow

Real data work is rarely a single clever line. It is a repeatable process:

1. **Load** the data into pandas.
2. **Inspect** it: shape, columns, types, a first look.
3. **Clean** it: missing values, duplicates, wrong types.
4. **Explore and aggregate**: group, pivot, summarise.
5. **Analyse**: compute the numbers that answer your question.
6. **Visualise and communicate**: charts and a clear conclusion.

This stage walks through all of it using financial data, but the skills (loading,
cleaning, grouping, summarising, plotting) are the everyday toolkit of any data analyst.

## The dataset

The file `data/stock_history.csv` holds three years of daily trading data for four
**fictional** companies (tickers `ATLAS`, `HELIOS`, `NORTH`, `VERTEX`). It is generated
with a fixed random seed by `generate_data.ipynb`, so it is reproducible and contains no real
market or personal data.

Each row is one ticker on one day, with these columns:

- `Date`, `Ticker`
- `Open`, `High`, `Low`, `Close` (prices)
- `Volume` (shares traded)

Let's load it and take a first look.

```python
import pandas as pd

df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"])
print(df.shape)        # (rows, columns)
df.head()
```

`parse_dates=["Date"]` tells pandas to read the Date column as real dates rather than
text, which matters as soon as you sort or resample by time.
