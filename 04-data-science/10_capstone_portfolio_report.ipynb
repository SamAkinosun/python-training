```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1-2: load, clean, reshape, returns
df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"]).drop_duplicates().sort_values("Date")
closes = df.pivot_table(index="Date", columns="Ticker", values="Close")
closes = closes.ffill()
returns = closes.pct_change().dropna()

# 3: portfolio
weights = pd.Series({"ATLAS": 0.4, "HELIOS": 0.2, "NORTH": 0.3, "VERTEX": 0.1})
portfolio_returns = returns.mul(weights, axis=1).sum(axis=1)

# 4: headline numbers
growth = (1 + portfolio_returns).cumprod()
total_return = (growth.iloc[-1] - 1) * 100
annual_vol = portfolio_returns.std() * np.sqrt(252) * 100
drawdown = (growth - growth.cummax()) / growth.cummax()
max_dd = drawdown.min() * 100

print(f"Total return:        {total_return:6.1f} %")
print(f"Annual volatility:   {annual_vol:6.1f} %")
print(f"Maximum drawdown:    {max_dd:6.1f} %")

# 5: correlation
print("\nReturn correlation:")
print(returns.corr().round(2))

# 6: growth chart
asset_growth = (1 + returns).cumprod()
ax = asset_growth.plot(figsize=(9, 5), alpha=0.5, title="Portfolio vs assets: growth of 1 unit")
growth.plot(ax=ax, label="PORTFOLIO", linewidth=2.5, color="black")
ax.set_ylabel("Growth factor")
ax.legend()
plt.tight_layout()
plt.show()
```
