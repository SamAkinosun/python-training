# Exploring and Aggregating

## What you'll learn
- Summarising groups with `groupby`
- Reshaping with `pivot_table`
- Resampling a time series to a coarser frequency

```python
import pandas as pd

df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"])
df = df.drop_duplicates().sort_values("Date")
df.head()
```

## Grouping

`groupby` answers "this per that". Here: average closing price and total volume per ticker.

```python
print(df.groupby("Ticker")["Close"].mean())
print("---")
print(df.groupby("Ticker")["Volume"].sum())
```

`agg` computes several summaries at once, which is how you build a quick comparison table.

```python
summary = df.groupby("Ticker")["Close"].agg(["min", "max", "mean", "std"]).round(2)
print(summary)
```

## Reshaping to wide form

The data is **long**: one row per ticker per day. For comparing tickers over time it is
easier in **wide** form: one row per date, one column per ticker. `pivot_table` does this.

```python
closes = df.pivot_table(index="Date", columns="Ticker", values="Close")
print(closes.head())
```

## Resampling a time series

With a date index, `resample` regroups by time period. `"ME"` means month-end. This turns
daily prices into a monthly average, which is far easier to scan than hundreds of daily rows.

```python
monthly = closes.resample("ME").mean().round(2)
print(monthly.head())
```

## Summary

- `groupby("col")["other"].agg([...])` summarises each group; great for comparisons.
- `pivot_table(index=, columns=, values=)` reshapes long data into a wide table.
- With a datetime index, `resample("ME")` / `"QE"` / `"YE"` regroups by month, quarter, year.
