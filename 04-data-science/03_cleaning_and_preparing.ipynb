# Cleaning and Preparing Data

## What you'll learn
- Removing duplicate rows
- Finding and handling missing values sensibly
- Why forward-fill suits price data
- Confirming the result is clean

```python
import pandas as pd

df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"])
print("rows:", len(df))
```

## Duplicates

Duplicate rows inflate counts and distort averages. `duplicated()` flags them and
`drop_duplicates()` removes them.

```python
print("duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print("after removing:", len(df))
```

## Missing values

`isna().sum()` counts gaps per column. This dataset has a few missing `Close` prices.

```python
print(df.isna().sum())
```

You have two broad choices: drop the affected rows, or fill the gaps. For a price series,
dropping a day breaks the timeline, so filling is usually better. **Forward fill**
(`ffill`) carries the last known price forward, which is a reasonable assumption for a
missing trading day. Fill within each ticker so one company's price never leaks into
another.

```python
df = df.sort_values(["Ticker", "Date"])
df["Close"] = df.groupby("Ticker")["Close"].ffill()
print("missing Close after fill:", df["Close"].isna().sum())
```

## Check the types

Make sure each column is the type you expect. `Date` should be a datetime (we parsed it on
load) and the price/volume columns numeric.

```python
print(df.dtypes)
```

## Confirm it is clean

A final check: no duplicates, no missing values, sensible shape. Cleaning is done when you
can state plainly that the data is ready.

```python
print("duplicates:", df.duplicated().sum())
print("missing values total:", df.isna().sum().sum())
print("shape:", df.shape)
```

## Summary

- `duplicated().sum()` counts duplicate rows; `drop_duplicates()` removes them.
- `isna().sum()` counts missing values per column.
- For price series, `ffill` (forward fill) is usually better than dropping rows; fill
  within each group so values do not leak across tickers.
- Finish by confirming no duplicates and no missing values remain.
