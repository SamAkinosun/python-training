# Visualising Financial Data

## What you'll learn
- Comparing prices fairly by rebasing them
- The shape of returns (a histogram)
- A correlation heatmap
- Choosing the right chart for the question

```python
import pandas as pd

df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"])
df = df.drop_duplicates().sort_values("Date")
# One price column per ticker, indexed by date:
closes = df.pivot_table(index="Date", columns="Ticker", values="Close")
closes = closes.ffill()   # carry the last known price over any gaps
print(closes.head())
```

## Rebased price comparison

Plotting raw prices together is misleading because they start at different levels. **Rebase**
each series to 100 at the start (divide by its first value, times 100) so you compare
percentage growth on equal footing.

```python
import matplotlib.pyplot as plt

rebased = closes / closes.iloc[0] * 100
ax = rebased.plot(figsize=(9, 5), title="Growth comparison (all start at 100)")
ax.set_ylabel("Rebased price")
ax.legend()
plt.tight_layout()
plt.show()
```

## The shape of returns

A histogram of daily returns shows how spread out they are. A wider spread means a riskier
asset. Compare a calm ticker with a volatile one.

```python
returns = closes.pct_change().dropna()

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(returns["NORTH"], bins=40, alpha=0.6, label="NORTH (calm)")
ax.hist(returns["HELIOS"], bins=40, alpha=0.6, label="HELIOS (volatile)")
ax.set_title("Distribution of daily returns")
ax.set_xlabel("Daily return")
ax.set_ylabel("Number of days")
ax.legend()
plt.tight_layout()
plt.show()
```

## Correlation heatmap

A correlation matrix is easier to read as a coloured grid. `imshow` draws it; we annotate
each cell with its value.

```python
corr = returns.corr()

fig, ax = plt.subplots(figsize=(5, 4))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
ax.set_xticks(range(len(corr)))
ax.set_xticklabels(corr.columns, rotation=45)
ax.set_yticks(range(len(corr)))
ax.set_yticklabels(corr.index)
for i in range(len(corr)):
    for j in range(len(corr)):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")
ax.set_title("Return correlation")
fig.colorbar(im)
plt.tight_layout()
plt.show()
```

## Choosing a chart

- **Line (rebased)**: compare growth of several assets fairly.
- **Histogram**: see the spread (risk) of returns.
- **Heatmap**: read a correlation matrix at a glance.
- **Bar**: compare a single summary value (total return, volatility) across assets.

Always title and label charts so they can be understood on their own.

## Practice Solutions

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"]).drop_duplicates().sort_values("Date")
closes = df.pivot_table(index="Date", columns="Ticker", values="Close").ffill()
returns = closes.pct_change().dropna()

# 1.
(closes / closes.iloc[0] * 100).plot(figsize=(9, 5), title="Rebased to 100")
plt.tight_layout()
plt.show()

# 2.
plt.figure()
plt.hist(returns["ATLAS"], bins=40)
plt.title("ATLAS daily returns")
plt.tight_layout()
plt.show()

# 3.
corr = returns.corr()
plt.figure()
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar()
plt.title("Correlation")
plt.tight_layout()
plt.show()

# 4.
total = ((1 + returns).cumprod().iloc[-1] - 1) * 100
total.plot(kind="bar", title="Total return %")
plt.tight_layout()
plt.show()
```
