# Financial Returns

## What you'll learn
- Why returns matter more than raw prices
- Daily returns with `pct_change`
- Cumulative returns (growth of an investment)
- Simple vs log returns

Prices on their own are hard to compare: a 1 dollar move means very different things for a
30 dollar stock and a 240 dollar stock. **Returns** (percentage changes) put everything on
the same footing. We start from clean, wide closing prices.

```python
import pandas as pd

df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"])
df = df.drop_duplicates().sort_values("Date")
# One price column per ticker, indexed by date:
closes = df.pivot_table(index="Date", columns="Ticker", values="Close")
closes = closes.ffill()   # carry the last known price over any gaps
print(closes.head())
```

## Daily returns

`pct_change()` gives the fractional change from one row to the next. The first row is `NaN`
because there is no prior day to compare to, so we drop it.

```python
returns = closes.pct_change().dropna()
print(returns.head())
```

## Cumulative returns

To see how an investment would have grown, compound the daily returns: add 1 to each,
multiply them cumulatively, and you get the growth factor over time. Multiply by 100 to read
it as "value of 100 invested at the start".

```python
growth = (1 + returns).cumprod()
print((growth * 100).round(1).tail())   # value today of 100 invested on day one
```

## Total return per ticker

The last row of the growth factor, minus 1, is the total return over the whole period.

```python
total_return = (growth.iloc[-1] - 1) * 100
print(total_return.round(1).sort_values(ascending=False))
```

## Simple vs log returns

The `pct_change` above gives **simple** returns. **Log** returns (`ln(price_today /
price_yesterday)`) are also common because they add up neatly over time. For small daily
moves the two are very close; you will see both in finance.

```python
import numpy as np

log_returns = np.log(closes / closes.shift(1)).dropna()
print(log_returns.head())
```

## Summary

- Returns make assets of different prices comparable.
- `pct_change()` gives daily simple returns; drop the first `NaN`.
- `(1 + returns).cumprod()` compounds them into cumulative growth.
- Final growth minus 1 is the total return.
- Log returns `np.log(p / p.shift(1))` are an alternative that adds over time.

## Practice Solutions

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"]).drop_duplicates().sort_values("Date")
closes = df.pivot_table(index="Date", columns="Ticker", values="Close").ffill()
returns = closes.pct_change().dropna()

# 2.
print(returns.mean())

# 3.
growth = (1 + returns).cumprod()
print((growth.iloc[-1] * 100).round(1))

# 4.
print("Highest total return:", (growth.iloc[-1] - 1).idxmax())

# 5.
log_returns = np.log(closes / closes.shift(1)).dropna()
print(log_returns.std())
```
