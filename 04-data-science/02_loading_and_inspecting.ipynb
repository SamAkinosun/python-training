# Loading and Inspecting Data

## What you'll learn
- The commands that tell you what a dataset contains
- Selecting one ticker to work with
- Sorting and setting a date index

```python
import pandas as pd

df = pd.read_csv("data/stock_history.csv", parse_dates=["Date"])
df.head()
```

## Know your data

Before analysing anything, answer: how big is it, what columns and types, and are there
gaps? These four commands cover it.

```python
print(df.shape)            # (rows, columns)
print(df.dtypes)           # the type of each column
```

```python
df.info()                  # types plus non-missing counts in one view
```

```python
df.describe()              # summary statistics for numeric columns
```

`head()`, `tail()`, and `sample()` show the first rows, last rows, and a random few. Use
them constantly to sanity-check your work.

```python
print(df.tail(3))
print("---")
print(df.sample(3, random_state=0))
```

## Focusing on one ticker

A filter keeps only the rows you want. Here we pull out a single company and sort it by
date so it reads as a time series.

```python
atlas = df[df["Ticker"] == "ATLAS"].sort_values("Date")
print(atlas.shape)
atlas.head()
```

## A date index

For time-based work it helps to make `Date` the index. Then rows are labelled by date and
time operations (which you will use in later lessons) become natural.

```python
atlas = atlas.set_index("Date")
print(atlas.loc["2021-01-04":"2021-01-08", ["Close", "Volume"]])
```

## Summary

- `shape`, `dtypes`, `info()`, and `describe()` tell you size, types, gaps, and statistics.
