### Pandas Series (1D Array)

A new data type available in Pandas, is the `series`, which is a one dimensional array holding any type of data. 

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0])
print(data)
```

For a full list of available functions of a `series`, look here: https://pandas.pydata.org/pandas-docs/version/0.23.4/api.html#series<br>

You can add labels (custom indexes) to each item in the `series`, similar to a `dictionary`, using a separate index list (note square brackets indicating the labels are actually presented as a `list`.

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
print(data)      # print the whole series
print(data['c']) # use square bracket to access a data item using its label
```

We can loop through a series to access one item at a time. Note below that each item in a series is returned as a `tuple`, which holds the data item and its label.

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=["a", "b", "c", "d"])
for value in data.items(): 
    print(value)
    print(type(value))
```

As we have seen previously with dictionaries, if you supply two variables to a `for` loop, you can get both the data item and the index (label):

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=["a", "b", "c", "d"])
for index, value in data.items():    # Ask for the data item and index (label)
    print(f"At index {index} the value is {value}")
```

As a Pandas `series` seems to be so similar to a `dictionary`, it is possible to convert a `dictionary` to a `series`:

```python
pop1831_dict = {"Dallas": 536853,
                "Minneapolis": 979000,
                "Chicago": 1292300,
                "Seattle": 536853}
population = pd.Series(pop1831_dict)  #convert dictionary to a series
print(population)
print(type(population))
```

## Pandas Dataframe (2D Array)

A `dataframe` is a two dimensional array, think of it like storing data in a spreadsheet.

You can create a `dataframe` from multiple `dictionaries` if each dictionary has the same index values. The index values will be converted to row names and the data items become columns. You can assign column names while `creating` the `dataframe`.

```python
pop1831_dict = {"Dallas": 536853,
                "Minneapolis": 979000,
                "Chicago": 1292300,
                "Seattle": 536853}

pop2019_dict = {"Dallas": 855800,
                "Minneapolis": 1846660,
                "Chicago": 907760,
                "Seattle": 761350}

population = pd.DataFrame({"1831": pop1831_dict,
"2019": pop2019_dict})
print(population)
```

If you have acquired data in separate `lists`, these can be joind together to form a `dataframe`, assuming of course that the lists are in the same order.:

```python
# Source https://www.kaggle.com/geomack/spotifyclassification?select=data.csv
song = ["In Flagranti", 
        "Baris K Edit", 
        "Cemalim", 
        "One Night"]

loudness = [-7.372, 
            -8.62, 
            -11.062, 
            -12.686]

df = pd.DataFrame({"Song":song, "Loudness":loudness})
print(df)
```

### Add / Remove a Coulmn

To Add a column to a `dataframe`

```python
df = pd.read_csv('populations.csv')
print(df,"\n")

# Create the new columns
new_col = [956800,1946360,1008760,864350]

# Insert the new column
df.insert(0,"Prediction 2026", new_col) # 0=col position
print(df)
```

Note how `insert()` has inserted the new column as the first column. This was due to `0` being passed in as the first parameter to `insert()`.To insert the column at the right side of the `dataframe` we need to work out how many columns there are:

```python
df = pd.read_csv('populations.csv')
print(df,"\n")

# Create the new columns
new_col = [956800,1946360,1008760,864350]

# Work out how many columns there are, and use this in insert()
end_col = len(df.columns)

# Insert the new column
df.insert(end_col,"Prediction 2026", new_col) # 0=col position
print(df)
```

To delete a column, use `drop()` as described here https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.drop.html 

### Add / Remove a Row

To add a row into a `dataframe`:

```python
df = pd.read_csv('populations.csv')
print(df,"\n")

# Create new row
new_row = [{"County": "Hertfordshire",
           "1831": 600654,
           "2019": 900343}]

# append new row to end of dataframe
df = df.append(new_row, ignore_index=True)
print(df)
```

To delete a row, use drop() as described here https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.drop.html 

### Access a single Column

To access a single column use square brackets with the column name. Note that the results are passed back as a series

```python
df = pd.read_csv('populations.csv')
print(df,"\n")

print(df["1831"],"\n")
print(type(df["1831"]))
```

### Access a single Row

To extract a single row, you can use `loc()` or `iloc()`, and supply a row index number. An overview of the differences between these two approaches is here https://towardsdatascience.com/how-to-use-loc-and-iloc-for-selecting-data-in-pandas-bd09cb4c3d79<br>
As a brief summary the format for `loc()` is:<br>
    `loc(row_label/s, column label/s)` <br>
If you want to select all rows or all columns, use a colon `:` for the row or columns. Take note of the location f the `:` inside `loc()` in the remaining examples below.

```python
df = pd.read_csv('populations.csv')
print(df,"\n")

print(df.loc[2,:],"\n")   # Supply row index nuumber
print(type(df.loc[2,:])) # only rows are being selected here, all columns will be returned
```

### Change the Index Column

If you need to use the row labels, you can change the column which is used as the index by using `set_index()`:

```python
df = pd.read_csv('populations.csv')
print(df,"\n")
df.set_index("County", inplace=True) # change the index column
print(df.loc["Norfolk",:])  # now use the values in county column as the index
```

### Access a Single Cell

```python
df = pd.read_csv('populations.csv')
print(df,"\n")
df.set_index("County", inplace=True) # change the index column
print(df.loc["Norfolk", "1831"])
```

### Selection Test

You can use basic evaluations when extracting data from a `dataframe`, using the comparison operators such as `>` `<=` etc

```python
df = pd.read_csv('populations.csv')
print(df,"\n")
df.set_index("County", inplace=True) # change the index column
print(df.loc[df ["2019"] > 860000,:])
```

### Slicing a Dataframe

You can also slice `dataframes` using the same approach as with strings, excepthere to slice whole rows out of the `dataframe`

```python
# SLICE ROWS

df = pd.read_csv('populations.csv')
print(df,"\n")
df.set_index("County", inplace=True) 
print(df.loc["Essex":"Suffolk",:])     # slice the dataframe by rows
```

```python
# SLICE COLUMNS

df = pd.read_csv('populations.csv')

# Insert new column at end of dataframe
new_col = [956800,1946360,1008760,864350]
end_col = len(df.columns)
df.insert(end_col,"2026", new_col) # 0=col position
print(df,"\n")


df.set_index("County", inplace=True) 
print(df.loc[:,"2019":"2026"])     # slice the dataframe by columns
```

### Transpose a Dataframe

Transposing is the process of rotating the data in a `dataframe` so that rows become columns and columns become rows.

```python
df = pd.read_csv('populations.csv')
print(df, "\n")
df_transp = df.transpose()
print(df_transp, "\n")
```

### Sort by Row (index)


```python
df = pd.read_csv('populations.csv')
df.set_index("County", inplace=True)  # set index col to 'county'
print(df,"\n")

# sort the rows in ascending order (index col is now 'county')
print(df.sort_index(axis=0, ascending=False),"\n")
```

### Sort All Rows by One Column (Values)

```python
df = pd.read_csv('populations.csv')
print(df,"\n")

# sort the columns
print(df.sort_values("1831", axis=0)) # sort by the 1831 clumn
```
