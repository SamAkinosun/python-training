## Working with Lists

```python
customers = [2737, "Susan", 8441, "Nick"]
print(customers[0]) # using the print() method to allow multiple outputs from this Jupyter cell
print(customers[2])
```

### Adding New Items to a List
You cannot add a new item in a list using square brackets method, try the code below to see what happens if you do:

```python
new_customers[3] = "Bob"
```

You must use the append() method to add a new item to the end of the list, as below:

```python
customers.append("Mary")
print(customers)
```

or use insert() to insert a new item at a specified location in the list:

```python
customers = [2737, "Susan", 8441, "Nick"]
print(customers)
customers.insert(2, "Alex") # list numbering starts at 0
print(customers)
```

### Slicing

Slicing a list works to a similar principle as slicing a string, using square brackets and a colon to specify the start and end positions of the slice, as below. Remember, in a slice, the item at the start location __IS__ included in the slice, but the item at the end location __IS NOT__ included in the slice, look closely at the example below and note which characters are included in the slice:

```python
letters = ["v", "y", "a", "j", "f", "f", "o", "b", "f", "w"]
sub_list = letters[2:7]
print(sub_list)
```

Slice from the beginning of the string:

```python
letters = ["v", "y", "a", "j", "f", "f", "o", "b", "f", "w"]
print(letters[:3])
```

Slice backwards from the end of a list:

```python
letters = ["v", "y", "a", "j", "f", "f", "o", "b", "f", "w"]
print(letters[-3:])
```

### Join Lists Together

```python
customers1 = ["Susan", "Alex", "Nick", "Mary"]
print(customers1)
customers2 = ["Bob", "John", "Susan", "Alice"]
print(customers2)
all_customers = customers1 + customers2
print(all_customers)
```

## Lists Inside Lists

So far we have used lists which hold individual data items at each index position in the list. It is possible to insert another list into a specific index location, thereby creating a list within a list.
Look at the example below, index positions 0, 2 and 3 each hold an individual string. If you look at index position 1, that data item is actually another list!

```python
test_list = ["word", [56, 45, 64], "house", "hotel"]
print(test_list)
```

Why is this useful? Consider the scenario that you want o hold information on blood donors. For each blood donor you need to record their county, town and age. Those three items of data can be placed into a list for each donor. that information for each donor can then be placed into another list which holds all the donors in one list variable, see the slide below.

The information for each donor is recorded in a list shown in the 6 lists in the list below. these lists are then placed inside another list `all_donors` so that all the donors information is contained within one variable. Note the commas separating each sublist within the main list, just in the same way you separate individual data items in a list.

```python
all_donors = [
["Suffolk", "Ipswich", 35],
["Suffolk", "Stowmarket", 42],
["Suffolk", "Ipswich", 45],
["Norfolk", "Thetford", 38],
["Suffolk", "Ipswich", 40],
["Essex", "Colchester", 56]
]
print(all_donors)
```

To access an individual item within a sub list, you use two sets of square brackets: `[position in main list][position in sub list]`

```python
all_donors = [
["Suffolk", "Ipswich", 35],    # main list item 0
["Suffolk", "Stowmarket", 42], # main list item 1
["Suffolk", "Ipswich", 45],    # main list item 2
["Norfolk", "Thetford", 38],   # main list item 3
["Suffolk", "Ipswich", 40],    # main list item 4
["Essex", "Colchester", 56]    # main list item 5
]
print(all_donors[3][1])
```

## Copying Lists

Copying lists in Python does not behave as you might expect.Take the two lists below. __Note:__ *You will need to run the cells below in the order they are presented for the examples to work correctly.*

```python
a = [2, 5]
print(a)
b = [1,6]
print(b)
```

Lets try making a copy of `a` and placing it in `b`

```python
b = a
```

Lets check both variables again.

```python
print(a)
print(b)
```

Great, it looks like a copy of `a` has been placed into `b`

If we change one of the values in `b`, then `a` shoudl remain unchanged

```python
b[0] = 9
print(a)
print(b)
```

Both `a` and `b` have changed! How is that possible? We only changed the value in `b`?

Here is an important point you MUST remember about copying lists. They are COPIED BY REFERENCE. What does that mean?
 In the example below we have two lists `a` and `b`.

If we use `b=a` you might expect the value from `a` to be copied into the variable `b`. No, that not what happens in Python, although that is how it works in some other programming languages. if the actual __value__ is copied into the other variable, that is called **PASS BY VALUE**, which is not what happens in Python.

In Python, when we say `b=a` what actually happens is that the variable `b` is made to point to the contents of variable `a`, so now both `a` and `b` are pointing to the same location in memory (or box as shown in the slides). The variable `b` has been redirected to point to `a`. 

In the slide below, note that the variable name `b` has now moved on top of box `a`, to show how this redirecting works. That is why, when you change a value in `a`, then `b` will change too because `b` is pointing to the same memory location as `a`. This is called `PASS BY REFERENCE`.

Why do it this way which seems very confusing? It's partly due to being more memory efficient, as you now only need one variable rather than two. For more information on `PASS BY REFERENCE` see [here](http://foobarnbaz.com/2012/07/08/understanding-python-variables/)

It is important to remember this in the future if you find yourself copying lists, it can be dangerous if you do not understand it. you could end up introducing wrong data into your analysis.

### Using `PASS BY VALUE' to Copy a List
To copy lists using the `PASS BY VALUE` approach so that both variable point to their own memory locations and are independent of each other, use a whole slice operation, as below:

```python
a = [2, 5]
print(a)
b = [1,6]
print(b)
b = a[:]   # Use a whole slice copy (no start or end positions provided)
b[0] = 9
print(a)
print(b)
```

# Sets

Another built in type of collection Python are Sets. Sets are a collection of __UNORDERED__, __UNINDEXED__ and __UNCHANGEABLE__ data. <br><br>
__DUPLICATE__ Data items are not allowed within a Set, and will be removed automatically by python if they are present in the Set.

To create a Set we use curly brackets with data items seperated by commas, as below. Note how duplicate entries are removed automatically.

```python
my_set = {"Blackbird", "Robin", "Blackbird", "Pheasant", "Robin"}
print(my_set)
```

You should be carefully when using sets. They can be useful to store data in a way that you cannot ever be changed, however, you will lose duplicate values which may result in inaccurate analysis.

Sets allow mathematical operations such as __union__, __intersection__, __difference__ and __symmetric difference__

Examples of these operations are shown below. You can see how sets can be useful when investigating similarities and differences between collections of data sets.

```python
obs1 = {"Blackbird", "Robin", "Blackbird", "Pheasant", "Robin"}
obs2 = {"Starling", "Wood Pigeon", "Robin", "Starling"}
print("observation 1 = ",obs1)
print("observation 2 = ",obs2,"\n")

# Set Intersection
print(f"Items common to both observations: { obs1 & obs2}\n")

# Set Union - !!returned as a unique set !!
print(f"All items from both observations (no duplicates): { obs1 | obs2}\n")

# Set Difference: Items unique to observation 1
print(f"Items unique to observation 1 when compared to observation 2: {obs1 - obs2}\n")

# Set Symmetric Difference
print(f"All items from both observations but not in both: { obs1 ^ obs2}\n")
```

## Creating an Empty Set

If you need to create an empty set use the `set()` function. Note that curly brackets are not used here as you are calling a function which does the work of creating the set for you.

```python
my_set = set()
my_set.add("Blue Tit")
print(my_set)
```
