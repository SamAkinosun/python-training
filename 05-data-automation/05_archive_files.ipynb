# Step 5: Archive the Processed Files

Once the daily files have been combined and processed, move them out of the incoming
folder into a dated archive folder. This keeps the incoming folder ready for the next
batch and means files are never processed twice.

Run it with **Run > Run All Cells**. It reads and writes files relative to this folder,
so keep the notebook in `05-data-automation/`.

```python
import os
import glob
import shutil
from datetime import datetime

BASE = os.getcwd()
INCOMING = os.path.join(BASE, "data", "incoming")
ARCHIVE = os.path.join(BASE, "data", "archive")
```

```python
def archive():
    files = sorted(glob.glob(os.path.join(INCOMING, "*.csv")))
    if not files:
        print("Nothing to archive: data/incoming/ is empty.")
        return []

    # A folder named for the run date, e.g. archive/2024-03-08/.
    run_date = datetime.now().strftime("%Y-%m-%d")
    target = os.path.join(ARCHIVE, run_date)
    os.makedirs(target, exist_ok=True)

    moved = []
    for path in files:
        name = os.path.basename(path)
        shutil.move(path, os.path.join(target, name))
        moved.append(name)
        print(f"  archived {name}")

    print(f"Moved {len(moved)} file(s) into archive/{run_date}/")
    print("(Run generate_sample_data.ipynb again to create a fresh batch.)")
    return moved
```

## Run the step

```python
moved = archive()
```
